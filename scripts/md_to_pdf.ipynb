## CONVERTS MARKDOWN TO PDF'S

```python
# ============================================================
# INSTALL REQUIRED PACKAGES
# ============================================================

import sys
import subprocess

packages = [
    "pypandoc",
    "pypandoc_binary",
    "playwright"
]

for package in packages:
    print(f"Installing/checking: {package}")
    subprocess.check_call([
        sys.executable,
        "-m",
        "pip",
        "install",
        "-U",
        package
    ])

print("\nPython packages installed.")
```

```python
# ============================================================
# INSTALL CHROMIUM FOR PDF RENDERING
# ============================================================

import sys
import subprocess

subprocess.check_call([
    sys.executable,
    "-m",
    "playwright",
    "install",
    "chromium"
])

print("\nChromium installation complete.")
```
